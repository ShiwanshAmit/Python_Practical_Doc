### Practical 3 - Dictionaries and Tuples

### Part A - Dictionaries

1. `Create an empty Dictionary.`

```python
inventory = {}
print(inventory)
```

2. `Store the first product details in variable:`
 * Product Name = Mobile Phone
 * Product Quantity = 5
 * Product Price = 20000
 * Product Release Year = 2020

```python
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_ReleaseYear = 2020

print(ProductNo1_Name)
print(ProductNo1_Quantity)
print(ProductNo1_Price)
print(ProductNo1_ReleaseYear)
```

3. `Add details in inventory`

```python
inventory[ProductNo1_Name] = {
    "Quantity": ProductNo1_Quantity,
    "Price": ProductNo1_Price,
    "ReleaseYear": ProductNo1_ReleaseYear
}

print(inventory)
```

4. `Store the Second Product details in variable`

```python
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 50000
ProductNo2_ReleaseYear = 2022

print(ProductNo2_Name)
print(ProductNo2_Quantity)
print(ProductNo2_Price)
print(ProductNo2_ReleaseYear)
```

5. `Add the item details in inventory`

```python
inventory[ProductNo2_Name] = {
    "Quantity": ProductNo2_Quantity,
    "Price": ProductNo2_Price,
    "ReleaseYear": ProductNo2_ReleaseYear
}

print(inventory)
```

6. `Display the products present in inventory`

```python
for product, details in inventory.items():
    print("Product:", product)
    print("Quantity:", details["Quantity"])
    print("Price:", details["Price"])
    print("Release Year:", details["ReleaseYear"])
    print()
```

7. `Check if ProductNo1_releaseYear and ProductNo2_releaseYear are in inventory.`

```python
if any(details["ReleaseYear"] == ProductNo1_ReleaseYear for details in inventory.values()):
    print("ProductNo1_ReleaseYear is present in inventory.")
else:
    print("ProductNo1_ReleaseYear is not present in inventory.")

if any(details["ReleaseYear"] == ProductNo2_ReleaseYear for details in inventory.values()):
    print("ProductNo2_ReleaseYear is present in inventory.")
else:
    print("ProductNo2_ReleaseYear is not present in inventory.")
```

8. `Delete release year of both the products from the inventory`

```python
del inventory[ProductNo1_Name]["ReleaseYear"]
del inventory[ProductNo2_Name]["ReleaseYear"]

print(inventory)
```

### Part B - Tuples

*You are given a tuple of product prices in a store. Perform the following operations using python:*

1. `Create a tuple called prices containing the following values: (250, 300, 150, 400, 100, 350, 200)`

```python
prices = (250, 300, 150, 400, 100, 350, 200)
print(prices)
```

2. `Find and print the highest and lowest price.`

```python
highest_price = max(prices)
lowest_price = min(prices)

print("Highest price:", highest_price)
print("Lowest price:", lowest_price)
```

3. `Calculate and print the total sum of all prices.`

```python
total = sum(prices)

print("Total sum:", total)
```

4. `Convert the tuple into sorted list(ascending order) and print it.`

```python
sorted_prices = sorted(prices)

print("Sorted prices:", sorted_prices)
```

5. `Try to modify an element in the tuple and explain what happens`

```python
prices[0] = 500
```

*`EXPLAINATION:` A tuple is immutable in Python. This means that once a tuple is created, its elements cannot be changed, added, or removed.*
