```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Student": ["David", "Samuel", "Terry", "Evan"],
    "Age": [27, 24, 22, 32],
    "Country": ["UK", "Canada", "China", "USA"],
    "Course": ["Python", "Data Structures", "Machine Learning", "Web Development"],
    "Marks": [85, 72, 89, 76]
}

df = pd.DataFrame(data)

print(df)
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aarav", "Diya", "Rohan", "Ananya", "Kabir",
             "Meera", "Arjun", "Isha", "Vivaan", "Sara"],
    "Age": [24, 35, 42, 29, 31, 38, 26, 45, 33, 22],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [2500, 4200, 1800, 3600, 5100,
                        2900, 3300, 4700, np.nan, 2200]
}

df = pd.DataFrame(data)


print("First 5 rows:")
print(df.head())


print("\nSummary Statistics:")
print(df.describe())


print("\nMissing Values:")
print(df.isnull().sum())


df["Purchase_Amount"] = df["Purchase_Amount"].fillna(
    df["Purchase_Amount"].mean()
)

print("\nDataset after handling missing values:")
print(df)
```

```python

print("Customers older than 30:")
print(df[df["Age"] > 30])


male_total = df.loc[
    df["Gender"] == "Male", "Purchase_Amount"
].sum()

print("\nTotal Purchase Amount for Male Customers:", male_total)


female_total = df.loc[
    df["Gender"] == "Female", "Purchase_Amount"
].sum()

print("Total Purchase Amount for Female Customers:", female_total)


print("\nSelected Columns:")
print(df[["Customer_ID", "Name", "Purchase_Amount"]])
```

```python

df["Discounted_Price"] = df["Purchase_Amount"] * 0.90

print("Dataset with Discounted Price:")
print(df)


print("\nAverage Purchase Amount by Gender:")
print(df.groupby("Gender")["Purchase_Amount"].mean())


sorted_df = df.sort_values(
    by="Purchase_Amount",
    ascending=False
)

print("\nSorted Dataset:")
print(sorted_df)
```
